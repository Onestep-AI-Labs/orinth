# Dataset EDA

Pick a dataset, then look at what is actually in it —
class balance first, because an imbalanced split is the single most
common reason a model trains and then scores badly.


```python
import orinth
import polars as pl

for dataset in orinth.datasets.list():
    print(f"{dataset.id:38} {dataset.task_type:20} {dataset.total_items:>7} items  {dataset.readiness.state}")
```

```python
DATASET_ID = orinth.datasets.list()[0].id

df = orinth.datasets.load(DATASET_ID, "train")
print(df.shape, df.columns)
df.head()
```

## Class balance

`readiness` reports an undrained inbox and a missing split, but not skew.
This is where you see it.


```python
label_col = "label" if "label" in df.columns else None
if label_col:
    counts = df.group_by(label_col).len().sort("len", descending=True)
    print(counts)
else:
    print("No single label column for this task — see the annotations column instead.")
```

## Text length

A long tail here is what silently truncates at training time.


```python
if "text" in df.columns:
    lengths = df.select(pl.col("text").str.len_chars().alias("chars"))
    print(lengths.describe())
```
